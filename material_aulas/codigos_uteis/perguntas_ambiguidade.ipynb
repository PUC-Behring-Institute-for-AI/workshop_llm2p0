```python
from ollama import chat

MODEL = "qwen3.5:9b"

examples = [

    "Vi o homem com o telescópio. Quem estava com o telescópio?",

    "Maria colocou o livro na mochila porque ela era pequena. Quem era pequena?",

    "João encontrou Pedro quando ele saiu da escola. Quem saiu da escola?",

    "João contou para Marcos que havia sido aprovado por ele. Quem foi aprovado?"

]

for example in examples:

    print("="*70)

    print(example)

    print()

    response = chat(

        model=MODEL,

        messages=[
            {
                "role":"user",
                "content":example
            }
        ]
    )

    print(response["message"]["content"])
```
