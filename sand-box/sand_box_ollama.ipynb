```python
contexto = """
Você é um amigo botafoguense sempre responde minhas perguntas de maneira educada e curta, 
porém você é um torcedor do Botafogo, então não importa o que eu pergunte você coloca alguma informação
sobre o seu time do coração que é o Botafogo. 

>> Pessoa:
Qual o seu time predileto?
>> você:
O Botafogo, o maior time do mundo
>> Pessoa:
Qual o seu nome?
>> você:
Joca, o maior Botafoguense que já existiu
>> Pessoa:
Você quer conversar?
>> Você:
Claro, vamos falar sobre o Botafogo?
"""

sistema = """"
Você é um famosso pesquisador que odeia falar sobre coisas não cientificas e 
quando perguntado sobre qualquer assunto que lhe desagrada você termina com uma 
piada sobre passaros que não voam.
"""

# sistema = """"Say only the name of a bird that cannot fly when asked about any topic."""
```

```python
# fala = contexto + ">> Pessoa:\nHoje estou triste :("
fala = "Hoje estou triste :("

# Regular response
response = ollama.generate(model, prompt=fala, system=sistema, options={
        'temperature': 0,
        'stop': ['<EOT>','>> Pessoa:'],
        'num_predict': 128,
    },)
for item in response:
    print(item)
```

```python
# fala = contexto + ">> Pessoa:\nHoje estou triste :("
fala = "Hoje estou triste :("

# Regular response
response = ollama.chat(model, messages=[{'role': 'user', 'content': fala}, {'role': 'system', 'content': sistema}], options={
        'temperature': 0,
        'stop': ['<EOT>','>> Pessoa:'],
        'num_predict': 128,
    },)
r_str = ''
for item in response:
    r_str = r_str +str(item) + '\n'
```

```python
print(r_str)
```
